```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import linear_model
import DataCleaning

df_clean = DataCleaning.df_clean
```

```python
df_clean["InvoiceMonth"] = pd.to_datetime(df_clean["InvoiceDate"]).dt.month #create a month variable for the linear Regression

# variables for the posttaxprofit
df_clean["PostTaxProfitIfNexus"] = df_clean["Profit"]-df_clean["SalesTaxNoNexus"] 
df_clean["ActualPostTaxProfit"] = df_clean["Profit"] - df_clean["SalesTaxOwed"]
# group dataframe for linear regression
df_grouped = df_clean.groupby(by=["State", "InvoiceYear", "InvoiceMonth"])[["ActualPostTaxProfit", "PostTaxProfitIfNexus"]].sum().reset_index()
df_grouped
growth_trends = []

# loop to predict the profit for all states
for state in df_grouped["State"].unique():
    df_state = df_grouped.loc[df_grouped["State"] == state].copy()
    
    # add the time index, one time unit equals one month unit
    df_state["time"] = range(len(df_state))
    
    #prediction time is set for in 6 months
    prediction_time = len(df_state) + 5

    #predict profit for the actual profit
    model_actual = linear_model.LinearRegression()
    model_actual.fit(df_state[["time"]], df_state[["ActualPostTaxProfit"]])
    actual_post_tax_profit_growth = model_actual.coef_[0][0]
    
    six_month_prediction_actual = model_actual.predict([[prediction_time]])[0][0]
    
    #predict profit for the profitif there was a nexus
    model_ifnexus = linear_model.LinearRegression()
    model_ifnexus.fit(df_state[["time"]], df_state[["PostTaxProfitIfNexus"]])
    post_tax_profit_ifnexus_growth = model_ifnexus.coef_[0][0]

    #get the forecasted profit according to the linear regression for december 2017
    last_actual_ifnexus = model_ifnexus.predict([[len(df_state) -1]])[0][0]

    #get the forecasted profit for in six months
    six_month_prediction_ifnexus = model_ifnexus.predict([[prediction_time]])[0][0]

    #calculate the absolute growth between these 6 months
    six_month_growth = six_month_prediction_ifnexus - last_actual_ifnexus

    #calculate the relative growth between these 6 months
    six_month_growth_percent = six_month_growth / last_actual_ifnexus * 100
    
    # save and append the predictions and the growth in the list
    growth_trends.append({
        "State": state,
        "ActualPostTaxProfit": six_month_prediction_actual,
        "ActualPostTaxProfitGrowth": actual_post_tax_profit_growth,
        "PostTaxProfitIfNexus": six_month_prediction_ifnexus,
        "PostTaxProfitIfNexusGrowth": post_tax_profit_ifnexus_growth,
        "LastPostTaxProfitIfNexus": last_actual_ifnexus,
        "SixMonthGrowth": six_month_growth,
        "SixMonthGrowthPercent": six_month_growth_percent
    })


# put list into dataframe
df_trends = pd.DataFrame(growth_trends)
df_trends = df_trends.sort_values(by=("PostTaxProfitIfNexus"), ascending=False)
df_trends


#create a multiple bar diagram to compare the profits under current and under nexus tax-scenarios
df_trends.plot(kind="bar",x="State", y=["ActualPostTaxProfit", "PostTaxProfitIfNexus"])

plt.xlabel("State", fontsize =12)
plt.ylabel("Post-Tax Profit", fontsize = 12)
plt.title("profit under current- and nexus-tax-scenario", fontsize =15)

#create a multiple bar diagram to compare the the profits in the six-month development 
df_trends.plot(kind="bar",x="State", y=["LastPostTaxProfitIfNexus", "PostTaxProfitIfNexus"])

plt.xlabel("State", fontsize =12)
plt.ylabel("Profit after Tax", fontsize = 12)
plt.title("Six-Month profit-development", fontsize =15)

# display the relative profit growth
df_trends = df_trends.sort_values(by=("SixMonthGrowthPercent"), ascending=False)
df_trends.plot(kind="bar",x="State", y="SixMonthGrowthPercent")

plt.xlabel("State", fontsize =12)
plt.ylabel("Relative Growth in %", fontsize = 12)
plt.title("Relative six-month profit-growth by state", fontsize =15)

                                             


```
